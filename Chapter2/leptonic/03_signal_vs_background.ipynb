# Chapter 2 (leptonic) — Notebook 3: Data vs stacked MC

**Goals**

- Produce ATLAS-style stacked-MC plots with **real data overlaid** and a Data/MC ratio panel.
- All MC histograms are filled with event weights.

```python
%matplotlib inline
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt

from topmass import io, kinematics, selection, plotting, fitting, neutrino, pairing, weights, style
from topmass.constants import M_W, M_TOP
```

## Worked example: stacked MET distribution with data overlay

```python
io.setup()
samples = io.build_samples()
procs = ['ttbar', 'single_top', 'diboson', 'data']
evs = {p: io.load_process(p, samples, fraction=0.1) for p in procs}
cuts = selection.SemilepCuts()

hists = {}
for p, ev in evs.items():
    sel = selection.semilep_preselection(ev, cuts)
    w = weights.weight_for(p, ev)[sel]
    hists[p] = plotting.make_hist(ev.met[sel], bins=40, range=(0, 200), weight=w, name='met')

plotting.stacked_plot(hists, xlabel='MET [GeV]')
```
