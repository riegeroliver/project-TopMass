# Chapter 2 (leptonic) — Notebook 1: Lepton selection

**Goals**

- Apply quality cuts on the lepton (pT, η, medium ID, loose isolation).
- Compare the lepton spectrum across signal and backgrounds (single-top, diboson) and real data.

```python
%matplotlib inline
import awkward as ak
import numpy as np
import matplotlib.pyplot as plt

from topmass import io, kinematics, selection, plotting, fitting, neutrino, pairing, weights, style
from topmass.constants import M_W, M_TOP
```

## Worked example: stream all processes

```python
io.setup()
samples = io.build_samples()
procs = ['ttbar', 'single_top', 'diboson', 'data']
evs = {p: io.load_process(p, samples, fraction=0.1) for p in procs}

cuts = selection.SemilepCuts()
for p, ev in evs.items():
    npass = int(ak.sum(selection.n_good_leptons(ev, cuts) == 1))
    print(f'{p:10s}  raw: {len(ev):7d}  exactly 1 good lepton: {npass:7d}')
```

```python
LEPTON_PT_MIN  = 25     # ✏️ try 25, 40, 60 GeV
LEPTON_ETA_MAX = 2.5    # ✏️ stretch: try 1.5

cuts = selection.SemilepCuts(lepton_pt_min=LEPTON_PT_MIN, lepton_eta_max=LEPTON_ETA_MAX)
for p, ev in evs.items():
    sel = selection.n_good_leptons(ev, cuts) == 1
    y = float(ak.sum(weights.weight_for(p, ev)[sel]))
    print(f'{p:10s}  weighted yield with 1 good lepton: {y:10.1f}')
```
